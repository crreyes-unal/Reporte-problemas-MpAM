***Probar que $log_2 P(X=i)$ es la cantidad de preguntas binarias que se necesitan para representar un mensaje con esa probabilidad. Realizar la demostración de por qué se puede expresar la entropía de una variable aleatoria mediante la fórmula adjunta***

Primero se probará que $log_2 P(X=i)$ es la cantidad de preguntas binarias que se necesitan para representar un mensaje con esa probabilidad.

Se  quiere encontrar la forma más eficiente para identificar un mensaje específico $x_i$ dentro de un conjunto de $n$ posibles mensajes. Este proceso se puede realizar mediante un árbol binario de decisión, la cantidad de preguntas necesarias para identificar el mensaje $x_i$ corresponde a la longitud del camino $l_i$ desde la raíz hasta la hoja del mensaje.

Buscamos minimizar el número promedio de preguntas $L$ dado por

$$L = \sum_{i=1}^n P(X=i) \cdot l_i$$

Para garantizar la unicidad y decodificación del árbol, las longitudes de las ramas $l_i$ deben satisfacer

$$\sum_{i=1}^n 2^{-l_i} \le 1$$

La condición de mayor eficiencia se tiene con la igualdad, esto es, con $\sum_{i=1}^n 2^{-l_i} = 1$.

Ahora se encuentra el valor optimo mediante multiplicadores de Lagrange. Para las longitudes $l_i$ se tiene la función

$$\mathcal{L}(l_1, \dots, l_n, \lambda) = \sum_{i=1}^n P(X=i) \cdot l_i + \lambda \left( \sum_{i=1}^n 2^{-l_i} - 1 \right)$$

Por criterio de la primera derivada se tiene

$$\frac{\partial \mathcal{L}}{\partial l_i} = P(X=i) + \lambda \frac{\partial}{\partial l_i} \left( 2^{-l_i} \right) = 0$$

Como $\frac{d}{dx}(2^{-x}) = - \ln(2) 2^{-x}$, entonces

$P(X=i) - \lambda \ln(2) 2^{-l_i} = 0 \implies P(X=i) = \lambda \ln(2) 2^{-l_i}$

Ahora,

$$\sum_{i=1}^n P(X=i) = \sum_{i=1}^n \lambda \ln(2) 2^{-l_i}$$

Dado que $\sum P(X=i) = 1$, y utilizando que $\sum 2^{-l_i} = 1$ se obtiene

$$1 = \lambda \ln(2) {\sum_{i=1}^n 2^{-l_i}}$$ la suma de le derecha da 1, así
$$\lambda = \frac{1}{\ln(2)}$$

Sustituyendo el valor de $\lambda$ en la anterior ecuación se obtiene

$$P(X=i) = \left( \frac{1}{\ln(2)} \right) \ln(2) 2^{-l_i}$$
$$P(X=i) = 2^{-l_i}$$

Por lo tanto

$$\log_2 P(X=i) = \log_2(2^{-l_i})$$
$$\log_2 P(X=i) = -l_i$$

$$l_i = -\log_2 P(X=i)$$

Por el tipo de problema, el número de preguntas $l_i$ debe ser entero, por lo que el número real óptimo de preguntas está acotado por

$$l_i^* = \lceil -\log_2 P(X=i) \rceil$$ tal como se quería.

Para la segunda pregunta se tiene que $H(X)$ se define como el valor esperado de la información contenida en una variable aleatoria discreta $X$.
Dado que la información de un evento individual $x_i$ es $I(x_i) = -\log_2 P(X=i)$ y dado que la entropía es la esperanza matemática de dicha información, esto esm $H(X) = \mathbb{E}[I(X)]$. Entonces, por definición de esperanza matemática para variables aleatorias discretas

$$H(X) = \sum_{i=1}^n P(X=i) \cdot I(x_i)$$

Sustituyendo $I(x_i) = -\log_2 P(X=i)$ en la sumatoria

$$H(X) = \sum_{i=1}^n P(X=i) \cdot \left( -\log_2 P(X=i) \right)$$

Por linealidad, se obtiene, tal como se quería

$$H(X) = -\sum_{i=1}^n P(X=i) \log_2 P(X=i)$$
