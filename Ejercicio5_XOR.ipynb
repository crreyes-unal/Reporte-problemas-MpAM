***Modelar un XOR conectando varias neuronas básicas. Luego modelar XOR con perceptrones (con pesos específicos para cada uno)***

```python
# Definir función de Heaviside
def activacion_paso(z):
    return 1 if z >= 0 else 0

# Definir perceptrones
def compuerta_or(x1, x2):
    # Pesos específicos y sesgo para OR
    w1, w2 = 1, 1
    b = -0.5
    z = (x1 * w1) + (x2 * w2) + b
    return activacion_paso(z)

def compuerta_nand(x1, x2):
    # Pesos específicos y sesgo para NAND
    w1, w2 = -1, -1
    b = 1.5
    z = (x1 * w1) + (x2 * w2) + b
    return activacion_paso(z)

def compuerta_and(x1, x2):
    # Pesos específicos y sesgo para AND
    w1, w2 = 1, 1
    b = -1.5
    z = (x1 * w1) + (x2 * w2) + b
    return activacion_paso(z)

# Modelar XOR
def red_xor(x1, x2):
    # Capa oculta
    h1 = compuerta_or(x1, x2)
    h2 = compuerta_nand(x1, x2)

    # Capa de salida
    y = compuerta_and(h1, h2)
    return y

#Verificación mediante tabla de verdad
entradas = [(0, 0), (0, 1), (1, 0), (1, 1)]

print(" Entrada X1 | Entrada X2 | Salida XOR ")
print("-" * 38)
for x1, x2 in entradas:
    resultado = red_xor(x1, x2)
    print(f"     {x1}     |     {x2}     |     {resultado} ")
```
