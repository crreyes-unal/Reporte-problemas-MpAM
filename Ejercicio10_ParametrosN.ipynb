***A partir de la desigualdad de Hoeffding determine, entre los parámetros $\epsilon$ & $\delta = 2Me^{-2N\epsilon^2}$ ¿cuál controla de una mejor manera el N? (N número de datos)***

Para determinar cuál parámetro controla de mejor manera el número de datos $N$, primero se despeja $N$ de la ecuación de la cota de unión de Hoeffding. Esto es

$$\delta = 2M e^{-2N\epsilon^2}$$
$$\frac{\delta}{2M} = e^{-2N\epsilon^2}$$
$$\ln\left(\frac{\delta}{2M}\right) = -2N\epsilon^2$$
$$\ln\left(\frac{2M}{\delta}\right) = 2N\epsilon^2$$
$$N = \frac{1}{2\epsilon^2} \ln\left(\frac{2M}{\delta}\right)$$

Se analizan ambos parámetros

*   El parámetro $\epsilon$ se encuentra elevado al cuadrado y en el denominador. Esto significa que si se quiere hacer la tolerancia a errores a la mitad, el número de datos $N$ necesarios se multiplicará por 4. Para reducir $\epsilon$ se necesita un crecimiento cuadrático del conjunto de datos.
*   El parámetro $\delta$ se encuentra dentro de un logaritmo natural. El logaritmo tiene un crecimiento lento. Si por ejemplo, se quiere que el modelo sea más seguro y reducir la probabilidad de error $\delta$ a una décima parte, el número de datos $N$ solo aumentará en una pequeña constante aditiva, específicamente $\ln(10) \approx 2.3$.

Así, pedirle al modelo reducir $\epsilon$ cuesta más datos que pedirle que reducir $\delta$. Por lo tanto, $\epsilon$ es el parámetro que controla de mejor manera el número de datos $N$. Adicional, se deduce que la precisión cuesta más datos que la confiabilidad.
