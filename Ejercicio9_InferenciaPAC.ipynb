***Si conocemos $\nu$ (fracción empírica) y $P[|\nu - \mu| \geq \epsilon] \leq 2e^{-2N\epsilon^2}$, decimos que $\nu = \mu$ es probablemente aproximadamente correcto (PAC).$\mu$ es la causa de $\nu$, pero podemos inferir que $\mu$ es aproximadamente igual que $\nu$. ¿Por qué?***

Primero, notemos que en la desigualdad de Hoeffding, se mide la distancia entre el valor empírico y el valor real que está dado por $|\nu - \mu|$.
Por las propiedades del valor absoluto, $|\nu - \mu| = |\mu - \nu|$. Esto dice que aunque la distribución real $\mu$ es la causa del resultado de la muestra $\nu$, matemáticamente no hay direccionalidad.

Ahora, antes de realizar un experimento la desigualdad de Hoeffding dice que si se conociera el $\mu$ real, es poco probable que una muestra aleatoria $\nu$ caiga lejos de él. Así mismo, por este resultado del valor absoluto se puede deducir que dado un valor $\nu$, es poco que $\mu$ esté lejos de lo que se acaba de medir.

Por lo tanto, se infiere que $\mu \approx \nu$ porque la desigualdad acota estrictamente la probabilidad de que la inferencia sea incorrecta.
A medida que el tamaño de la muestra $N$ crece, el término de error $2e^{-2N\epsilon^2}$ tiende hacia cero. Por tanto,es aproximadamente correcto pues que $\mu$ está en el intervalo $[\nu - \epsilon, \nu + \epsilon]$ . Y es probablemente cierto porque se afirma que esto ocurre con una confianza de al menos $1 - 2e^{-2N\epsilon^2}$.
